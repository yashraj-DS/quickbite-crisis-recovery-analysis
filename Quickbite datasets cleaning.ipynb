```python
from sqlalchemy import create_engine,text
```

```python
engine = create_engine(
    "mysql+pymysql://root:yash@localhost/analysis")
```

```python
conn = engine.connect()
```

```python
tables = {
    "dim_customer": dim_customer,
    "dim_delivery_partner": dim_delivery_partner,
    "dim_menu_item": dim_menu_item,
    "dim_restaurant": dim_restaurant,
    "fact_delivery_performance": fact_delivery_performance.
    "fact_order_items": fact_order_items,
    "fact_orders": fact_orders,
    "fact_ratings": fact_ratings}

for table_name,df in tables.items():
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists='replace',
        index=False,
        chunksize=30000,
        method='multi'
    )
    print(f"{table_name} imported")
```
